# ANÁLISIS EXPLORATORIO DE DATOS (EDA) Y PREPARACIÓN DE LA INFORMACIÓN PARA EL ESTUDIO DEL COMPORTAMIENTO DE USUARIOS DE NETFLIX

## Resumen

El presente trabajo tiene como objetivo analizar el comportamiento de los usuarios de una plataforma de streaming mediante técnicas de análisis exploratorio de datos (Exploratory Data Analysis, EDA). Para ello se integraron diferentes conjuntos de datos relacionados con usuarios, contenidos audiovisuales e interacciones de consumo, con el propósito de identificar patrones de comportamiento, preferencias de visualización y factores potencialmente asociados a la retención o abandono de clientes.

El estudio se desarrolló utilizando Python y diversas librerías especializadas en análisis de datos y visualización, permitiendo transformar los datos originales en información útil para la toma de decisiones estratégicas.

## 1. Introducción

La industria del streaming se caracteriza por una elevada competencia y una creciente necesidad de comprender el comportamiento de los usuarios para optimizar la experiencia de consumo y maximizar la retención de clientes. En este contexto, el análisis de datos constituye una herramienta fundamental para identificar hábitos de visualización, preferencias de contenido y patrones de uso que permitan diseñar estrategias más eficaces de personalización y fidelización.

El presente proyecto se centra en el análisis de una base de datos simulada inspirada en una plataforma de streaming similar a Netflix. A través de un proceso sistemático de exploración, limpieza, transformación e integración de datos, se pretende obtener una visión global del comportamiento de los usuarios y generar indicadores relevantes para la toma de decisiones.

## 2. Objetivos

### Objetivo general

Analizar el comportamiento de los usuarios de la plataforma mediante técnicas de análisis exploratorio de datos para identificar patrones de consumo, segmentación de clientes y factores relacionados con el abandono del servicio.


### Objetivos específicos

* Evaluar la calidad y estructura de los datos disponibles.
* Integrar múltiples fuentes de información mediante identificadores comunes.
* Analizar variables demográficas y de comportamiento.
* Identificar relaciones entre consumo de contenido y abandono.
* Detectar segmentos de usuarios con características diferenciadas.
* Generar una base analítica para la construcción de un dashboard ejecutivo.

## 3. Metodología

La metodología aplicada se estructuró en cinco fases principales:

### Fase 1. Comprensión de los datos

Se realizó una revisión inicial de los datasets para identificar:

* Número de registros y variables.
* Tipología de los datos.
* Variables numéricas y categóricas.
* Posibles inconsistencias o valores faltantes.

### Fase 2. Limpieza y preparación

Durante esta etapa se llevaron a cabo tareas de depuración y transformación de los datos, incluyendo:

* Verificación de registros duplicados.
* Revisión de valores nulos.
* Estandarización de formatos.
* Eliminación de variables redundantes.
* Creación de nuevas categorías para facilitar el análisis.

### Fase 3. Integración de fuentes

Los distintos datasets fueron combinados utilizando los identificadores:

* user_id
* content_id

Este proceso permitió construir una tabla analítica única que integra información de usuarios, contenidos e interacciones.

### Fase 4. Análisis exploratorio


Se desarrollaron análisis univariados y bivariados para estudiar tanto la distribución individual de las variables como las relaciones existentes entre ellas.

### Fase 5. Interpretación de resultados

Los hallazgos obtenidos fueron interpretados desde una perspectiva de negocio con el objetivo de identificar oportunidades de mejora en la gestión de usuarios y contenidos.

## 5. Análisis Univariado

El análisis univariado permitió comprender la distribución y comportamiento individual de cada variable relevante.

### Distribución de usuarios por edad

La variable edad fue segmentada en grupos generacionales con el objetivo de facilitar la interpretación de los resultados y detectar posibles diferencias de comportamiento entre segmentos demográficos.

### Tiempo de visualización

El análisis de los minutos visualizados permitió identificar distintos niveles de consumo y evaluar el grado de compromiso de los usuarios con la plataforma.

### Nivel de finalización de contenido

Esta variable se estudió para medir la capacidad de los contenidos para mantener el interés de la audiencia hasta su finalización.

### Frecuencia de acceso

La variable sesiones por semana permitió evaluar la intensidad de uso de la plataforma y distinguir entre usuarios ocasionales y recurrentes.

### Tipo de suscripción

Se examinó la distribución de usuarios según el plan contratado para comprender la estructura de ingresos de la plataforma.

### Dispositivos de acceso

El análisis de dispositivos proporcionó información sobre las preferencias tecnológicas de los usuarios.

### Distribución geográfica

Se analizaron los países presentes en la muestra para identificar la diversidad geográfica de la base de usuarios.

### Abandono del servicio

El estudio de la variable churn permitió cuantificar el porcentaje de usuarios que abandonan la plataforma.

## 6. Análisis Bivariado


El análisis bivariado tuvo como finalidad identificar relaciones entre variables relevantes para el negocio.

### Tiempo de visualización y abandono

Los resultados sugieren que los usuarios con menores niveles de consumo presentan una mayor probabilidad de abandonar la plataforma, lo que refuerza la importancia del engagement como indicador de retención.

### Tipo de suscripción y abandono

El análisis permitió detectar posibles diferencias en las tasas de cancelación entre los distintos planes disponibles.

### Interacción con recomendaciones y churn

La relación observada evidencia la relevancia de los sistemas de recomendación como herramienta para incrementar la participación del usuario y reducir el abandono.

### Edad y tipo de suscripción

Este análisis permitió identificar preferencias diferenciadas entre grupos etarios respecto a las modalidades de contratación.

### Consumo por género favorito

Los géneros con mayores niveles de visualización muestran una mayor capacidad para generar compromiso y fidelización.

### Ingresos asociados a preferencias de contenido

La comparación entre géneros permitió evaluar su contribución relativa al valor económico generado por la plataforma.

## 7. Justificación de las visualizaciones empleadas

La selección de visualizaciones respondió a criterios de claridad, capacidad comparativa y facilidad de interpretación.

Los gráficos de barras fueron utilizados para comparar frecuencias y distribuciones categóricas, facilitando la identificación de diferencias entre grupos.

Los histogramas permitieron analizar distribuciones de variables numéricas y detectar posibles concentraciones o dispersiones.

Los diagramas de dispersión se emplearon para estudiar relaciones entre variables cuantitativas y explorar posibles correlaciones.

Los mapas de calor facilitaron la identificación de patrones de asociación entre variables categóricas complejas.

La combinación de estas visualizaciones permitió obtener una comprensión integral del comportamiento de los usuarios.











## 8. Principales hallazgos

Los resultados obtenidos muestran que:

* El nivel de consumo constituye uno de los factores más relacionados con la permanencia en la plataforma.
* Existen diferencias significativas entre grupos demográficos respecto a hábitos de visualización y contratación.
* La interacción con recomendaciones parece influir positivamente en la retención.
* Determinados géneros de contenido generan mayores niveles de engagement.
* Los dispositivos utilizados reflejan patrones diferenciados de acceso y frecuencia de consumo.

## 9. Conclusiones ejecutivas

El análisis exploratorio permitió transformar datos dispersos en conocimiento accionable para la toma de decisiones empresariales.

Los hallazgos sugieren que las estrategias de retención deberían centrarse especialmente en usuarios con bajo nivel de consumo y escasa interacción con las recomendaciones de contenido. Asimismo, la segmentación por edad, país y preferencias de visualización ofrece oportunidades para desarrollar campañas de personalización más efectivas.
